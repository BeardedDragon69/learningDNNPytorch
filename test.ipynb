```python
#  testing the partial derivative of a particular function

x = torch.tensor(2.0, requires_grad= True)
y = torch.tensor(3.0, requires_grad= True)
z = torch.tensor(4.0, requires_grad= True)

f = x**3+ y**2 + z**2

f.backward()
x.grad, y.grad, z.grad
```

```python
inDim = 784
featureDim = 256
outDim = 10

vec = torch.randn(784, device=device)

layer1 = nn.Linear(inDim, featureDim,device=device, bias=True)
layer2 = nn.Linear(featureDim, outDim, device=device, bias=True)

out = (layer2(layer1(vec)))

print(out)
print(vec.shape)
# print(featureDim.shape)
```

```python
# non linear neuron

relu = nn.ReLU()

out = layer2(relu(layer1(vec)))

print(out)
```
