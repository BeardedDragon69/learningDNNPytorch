```python
import torch
import torch.nn as nn
```

```python
device = 'cuda'

inDim, featureDim, outDim = 784, 256, 10

vec = torch.randn(784)
```

```python
class BaseClassifier(nn.Module):
       def __init__(self, inDim, featureDim, outDim):
              super(BaseClassifier, self).__init__()
              self.layer1 = nn.Linear(inDim, featureDim, bias = True, device=device)
              self.layer2 = nn.Linear(featureDim, outDim, bias=True, device=device)

              self.relu = nn.ReLU().to(device=device)

       def forward(self, x):
              y = self.layer1(x)
              z = self.relu(y)
              out = self.layer2(z)

              return out
```

```python
noExamples = 10
x = torch.randn(noExamples, inDim, device=device)
classifier = BaseClassifier(inDim, featureDim, outDim)

out = classifier(x)

print(out)
```

```python
loss = nn.CrossEntropyLoss()
target = torch.tensor([0,3,2,8,2,9,3,7,1,6], device= device)
computedLoss = loss(out, target)
computedLoss.backward()
```
